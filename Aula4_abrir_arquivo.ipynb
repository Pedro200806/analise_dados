```python
import pandas as pd
import openpyxl

dado = r"CD2022_Populacao_2010_Compatibilizada_20231222.xlsx"

indicador_1 = pd.read_excel(dado, header=2)

indicador_1 = indicador_1.iloc[:,1:]

indicador_1 = indicador_1.loc[:,[
    "UF",
    "COD. UF",
    "COD. MUNIC",
    "NOME DO MUNICÍPIO",
    "População Município 2010\n(Sinopse)",
    "População 2010 (Alterações de Limites até 2022)1",
    "População Censo 2022"
]]

indicador_1.head()
```

```python
populacao_estado = (
    indicador_1.loc[:, [
        "UF",
        "População 2010 (Alterações de Limites até 2022)1",
        "População Censo 2022"
    ]]
    .groupby("UF")[[
        "População 2010 (Alterações de Limites até 2022)1",
        "População Censo 2022"
    ]]
    .sum()
    .reset_index()
)

# Calcular o crescimento
populacao_estado["Crescimento"] = (
    populacao_estado["População Censo 2022"] - populacao_estado["População 2010 (Alterações de Limites até 2022)1"]
)

# Organizar do maior para o menor crescimento
populacao_estado = populacao_estado.sort_values(
    "Crescimento",
    ascending=False
).reset_index(drop=True)

populacao_estado
```

```python
populacao_estado.to_csv(
    "populacao_por_estado_crescimento.csv",
    sep=";",
    index=False
)
```

```python
populacao_municipio = indicador_1.loc[:, [
    "UF",
    "COD. UF",
    "COD. MUNIC",
    "NOME DO MUNICÍPIO",
    "População 2010 (Alterações de Limites até 2022)1",
    "População Censo 2022"
]].copy()

# Calcular o crescimento
populacao_municipio["Crescimento"] = (
    populacao_municipio["População Censo 2022"]
    - populacao_municipio["População 2010 (Alterações de Limites até 2022)1"]
)

# Organizar do maior para o menor crescimento
populacao_municipio = populacao_municipio.sort_values(
    "Crescimento",
    ascending=False
).reset_index(drop=True)

populacao_municipio
```

```python
populacao_municipio.to_csv(
    "populacao_por_municipio_crescimento.csv",
    sep=";",
    index=False
)
```
